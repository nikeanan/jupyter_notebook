```python
import pandas as pd
```

```python
# read csv file
df = pd.read_csv('soil_geotech_data.csv')
# select columns
df = df[['Moisture','Crop Type','Nitrogen','Potassium','Phosphorous']]
```

```python
# show dataframe
df
```

# 1. Selecting with a single value
loc[row_label,column_label]

```python
# get all data about index 55
df.loc[55]
```

```python
df.loc[200]
```

```python
# get the Moisture of 75
df.loc[75,'Moisture']
```

```python
# get the phosphorous of 89 index
df.loc[89,'Phosphorous']
```

```python
# get all rows inside the 'Nitrogen' column
df.loc[:,'Nitrogen']
```

```python
# get all columns that correspond to the index 'loamy'
df.loc[45,:]
```

# 2. Selecting with a list of values

```python
# get all data about 200 and 20
df.loc[[200,20]]
```

```python
# get the Crop Type of 99 and 56
df.loc[[99,56],'Crop Type']
```

```python
# get the Phosphorous of 99 and 56
df.loc[[99,56],'Phosphorous']
```

```python
# get the Phosphorous and Nitrogen of 99
df.loc[[99],['Phosphorous','Nitrogen']]
```

```python
# get the Phosphorous and Nitrogen of 99 and 56
df.loc[[99,56],['Phosphorous','Nitrogen']]
```

# 3. Selecting a range of data with a slice
start:stop:step(Note that contrary to usual pythonslices,both the start and stop are included )

```python
# slice column labels
Crop_Type = [56,62]
df.loc[Crop_Type,'Nitrogen':'Phosphorous']
```

```python
# slice index labels
columns = ['Moisture','Nitrogen','Potassium']
# get top1 and top10 index
df.index[:10]

df.loc[0:10, columns]
```

# 4. Selecting with conditions

```python
# one condition: select player with Moisture below 23
columns =['Nitrogen','Moisture','Potassium','Phosphorous']

df.loc[df['Moisture']<23,columns]
```

```python
# multiple conditions: select Moisture <23  but Potassium>15
columns =['Nitrogen','Moisture','Potassium','Phosphorous']

df =df.loc[(df['Moisture']<23) & (df['Potassium']>15), :]
```
