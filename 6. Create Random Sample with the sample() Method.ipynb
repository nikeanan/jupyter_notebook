```python
import pandas as pd
```

```python
# read csv file
df = pd.read_csv('soil_geotech_data.csv')
# select columns
df = df[['Moisture','Crop Type','Nitrogen','Potassium','Phosphorous']]
```

```python
# show dataframe
df
```

# 1. sample()

```python
# extract 10 random elements from the "nationality" column
df['Crop Type'].sample(10, random_state=9)
```

```python
# extract a random 20% sample of the dataframe , new values everytime we run
df.sample(frac=0.2)
```

```python
df.sample(frac=0.05, random_state=20)
```

```python
# upsample: increase the sampling rate -> frac>1 (Note: replace  parameter has to be True for frac parameter > 1)

df.sample(frac=2,replace=True, random_state=99)
```
