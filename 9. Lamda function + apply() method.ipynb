# 2. Apply + Lambda Function

```python
# use lambda function to convert "Temperature in  degree Celscius " to degree Fahrenheit
df['Temparature'].apply(lambda x:1.8*x+32)
```

```python
# alternative
df['Temparature']*(1.8)+32
```

```python
# use lambda function to convert "Crop Type" series to upper case
df['Crop Type'].apply(lambda x:x.upper())
```

```python
# alternative with str attribute
df['Crop Type'].str.upper()
```

```python
df['Temp'] = df['Temparature'].astype('int64')
# use lambda function to get the before decimal digit of the Moisture
df['Temp'].apply(lambda x:x)
# alternative with str attribute
```

```python
def suggest_fertiliser(row):
    if row['Nitrogen'] > 30 and row['Phosphorous'] < 10:
        return 'urea'
    elif row['Phosphorous'] > 20 and row['Potassium'] < 10:
        return 'DAP'
    elif abs(row['Nitrogen'] - row['Phosphorous']) < 5:
        return '28-28'
    else:
        return '10-26-26'

df['suggested_fertiliser'] = df.apply(suggest_fertiliser, axis=1)
```
