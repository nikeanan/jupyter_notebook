```python
import pandas as pd
```

```python
# read csv file
df = pd.read_csv('soil_geotech_data.csv')
# select columns
df = df[['Moisture','Crop Type','Nitrogen','Potassium','Phosphorous']]
```

```python
# show dataframe
df
```

# 1. Drop Row

```python
# drop new row
# axis paramete
df.drop(1, axis=0)
```

```python
# index parameter
df.drop(index=[2])
```

```python
# drop two or more row and update data (inplace=True)
df.drop([2,4], axis=0, inplace= True)
```

```python
df
```

# 2. Drop columns

```python
# drop one column
# axis parameter
df.drop('Moisture', axis=1)
```

```python
# column parameter
df.drop(columns=['Moisture'])
```

```python
df
```

```python
# drop column by position/index (drop last column)

df.drop(df.columns[[-1]],axis=1)
```

```python
# drop two or more columns and update data (inplace=True)
df.drop(['Potassium','Nitrogen'], axis=1, inplace=True)
```
