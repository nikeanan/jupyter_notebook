```python
import pandas as pd
```

```python
# reading the csv file
df_exams = pd.read_csv('rainfall_by_districts_2019.csv')
df_exams
```

```python
df_exams['average'] = (df_exams["Actual Rainfall in South West Monsoon (June'17 to September'17) in mm"]+	df_exams["Normal Rainfall in South West Monsoon (June'17 to September'17) in mm"]+	df_exams["Actual Rainfall in North East Monsoon (October'17 to December'17) in mm"])
df_exams
```

# 1 sort a dataframe

```python
# sort by one column
df_exams.sort_values('average')
```

```python
# sort descending by one column
df_exams.sort_values('average', ascending=False)
```

```python
# sort descending by multiple columns
df_exams.sort_values(['average', "Total Normal Rainfall (June'17 to May'18) in mm"], ascending=False)
```

```python
# sort descending by multiple columns average as first priority and Total Normal Rainfall (June'17 to May'18) in mm as second priority
df_exams.sort_values(['average', "Total Normal Rainfall (June'17 to May'18) in mm"], ascending=False)
```

```python
# sort descending by multiple columns and update dataframe
df_exams.sort_values(['average', "Total Normal Rainfall (June'17 to May'18) in mm"], ascending=False,
                    inplace=True)
```

```python
# sort descending with a key function
df_exams.sort_values('District', ascending=False,
                    key=lambda col:col.str.lower())
```
