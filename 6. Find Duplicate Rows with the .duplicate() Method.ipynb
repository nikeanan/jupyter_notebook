```python
# find duplicate in 2 column/series
df_cpcb_dataset.duplicated('slag')
```

# 1.2 Example 2
Task: Find the etremely high strength and highest cement content laptop of ech company using the sort_values() and duplicated() method(keep "first", "last" and false)

```python
# sort dataframe ascending by cement and strength 
# least strength first and most strength last
df_cpcb_dataset = df_cpcb_dataset.sort_values(['cement','strength'])
df_cpcb_dataset
```

```python
# check all value categories in "age" column
df_cpcb_dataset.value_counts('age')
```

```python
# duplicate values in "age columns (default -> keep='first')
df_cpcb_dataset.duplicated('age', keep = 'first')
```

```python
# show dataframe with duplicated values in "age" column
duplicated_first = df_cpcb_dataset.duplicated('age', keep = 'first')
df_cpcb_dataset[duplicated_first]
# keep='first' (least strength per cement content)
# show dataframe with non duplicated values in "cement" column
df_cpcb_dataset[~duplicated_first][['cement','age']]
# check all value categories
df_cpcb_dataset[~duplicated_first].value_counts('age')
```

```python
# show dataframe with duplicated values in "cement" column
duplicated_first = df_cpcb_dataset.duplicated('cement', keep = 'first')
df_cpcb_dataset[duplicated_first]
# keep='first' (least strength per cement content)
# show dataframe with non duplicated values in "cement" column
df_cpcb_dataset[~duplicated_first][['cement','age']]
# check all value categories
df_cpcb_dataset[~duplicated_first].value_counts('cement')
```

```python
# keep='last'
duplicated_last = df_cpcb_dataset.duplicated('cement', keep ='last')
#keep=false
duplicated_false = df_cpcb_dataset.duplicated('cement', keep =False)
```

```python
# keep='last' (highest strength  per  cement content)
# show dataframe with non duplicated values in "cement" column
df_cpcb_dataset[~duplicated_last][['cement','strength']]
```

```python
# keep=false
# show dataframe with non duplicated values in "cement" column
df_cpcb_dataset[~duplicated_false]
```

```python
# all elements in cement column are repeated once
df_cpcb_dataset[~duplicated_false].value_counts('cement')
```

```python
# all elements in cement column are repeated at least once
df_cpcb_dataset.value_counts('cement')
```
