```python
import pandas as pd
```

```python
# read csv file
df = pd.read_csv('soil_geotech_data.csv')
# select columns
df = df[['Moisture','Crop Type','Nitrogen','Potassium','Phosphorous']]
```

```python
df.iloc[45,3]
```

```python
# get the Nitrogen of 102
df.loc[[102],'Nitrogen']
```

```python
df.iloc[102,2]
```

```python
# get all rows inside the 'Nitrogen'
df.loc[:,'Nitrogen']
```

```python
df.iloc[:,2]
```

```python
df.iloc[190,:]
```

# 2. Selecting with a list of values

```python
# get the Moisture of 4 and 17
df.iloc[[4,17],0]
```

```python
# get the Moisture and Potassium of 4 
df.iloc[[4],[0,2]]
```

# 3. Selecting a range of data with a slice
start:stop(Note the the start included but the stop is excluded)

```python
df
```

```python
# slice column labels Moisture/Nitrogen
# index = [4,17]
index = [4,17]

df.iloc[index,0:2+1]
```

```python
# slice index labels (top1 to top10 index)
# columns = ['Nitrogen','Potassium','Phosphorous']
index1 = [2,3,4]

# get top1 and top10 index
df.iloc[0:10,index1]
```

```python
df.iloc[0:10,2:4+1]
```

# 4. Selecting with conditions
iloc cannot accept a boolean Series, but only a boolean list. We have to use the list() function to convert a series into a boolean list.

```python
# one condition: select index with  phosphorous above 20
# columns = ['Moisture','Crop Type','Nitrogen']
columns = [1,2,3,4]

df.iloc[list(df['Phosphorous']>20), columns]
```

```python
# multiple conditions : select index with Phosphorous above 40 and potassium == 0

df.iloc[list((df['Phosphorous']>40) & (df['Potassium']==0)), columns]
```
