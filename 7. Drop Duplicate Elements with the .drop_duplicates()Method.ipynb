# 1 drop_duplicates()

```python
# drop duplicates in 2 or more columns
df_cpcb_dataset.drop_duplicates(['coarseagg'])[['coarseagg','fineagg']].value_counts('coarseagg')
```

```python
# sort dataframe ascending by coarseagg & fineagg
# least first and most last
df_cpcb_dataset = df_cpcb_dataset.sort_values(['coarseagg','fineagg'])
```

```python
# least: keep=='first'
df_cpcb_dataset.drop_duplicates(['coarseagg'], keep='first')[['coarseagg','fineagg']]
```

```python
# most expensive : keep='last'
df_cpcb_dataset.drop_duplicates(['coarseagg'], keep='last', inplace=True, ignore_index=True)
```

```python
# Arguments
# inplace: drop duplicates in place or to return a copy.
#ignore_index=True: the resulting
df_cpcb_dataset[['coarseagg','fineagg']]
```

# 2 Excercise 1
Task: Find the biggest and smallest age in cement of each strength
using the sort_values() and duplicated() methods (keep "first","last" and False)

```python
df_cpcb_dataset = pd.read_csv('concrete.csv')
```

```python
# sort dataframe ascending by age and strength
# smallest first and biggest last (by default)
df_cpcb_dataset.sort_values(['age','strength'])
```

```python
# smallest : keep='first'
df_cpcb_dataset.drop_duplicates(['age'], keep='first')[['age','strength']]
```

```python
# biggest : keep='last'
df_cpcb_dataset.drop_duplicates(['age'], keep='last')[['age','strength']]
```

```python
df_cpcb_dataset.drop_duplicates(['age'], keep=False)[['age','strength']]
```
