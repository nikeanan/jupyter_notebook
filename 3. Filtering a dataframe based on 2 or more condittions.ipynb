```python
import pandas as pd
```

```python
df_cpcb_dataset = pd.read_csv('concrete.csv')
```

```python
df_cpcb_dataset.head(4)
```

# 1. Filter dataframe based on a condition

```python
# find 0.0 superplastic concrete
df_cpcb_dataset['superplastic'] == 0.0
```

```python
# find concrete that have superplasticity more than or equal to 7
df_cpcb_dataset['superplastic'] >= 7.0
```

# 2. Filter dataframe based on multiple conditions

```python
# find cement 250 that have superplastic more than  or equal to 7.0
(df_cpcb_dataset['cement']== 250) & (df_cpcb_dataset['superplastic'] >= 7.0)
```

```python
and -> &
or -> |
```

# 2.1 & and

```python
# filter dataframe based on multiple conditions
df_cpcb_dataset[(df_cpcb_dataset['cement']== 250) & (df_cpcb_dataset['superplastic'] >= 7.0)]
```

# 2.2 | or

```python
# find Cement 250 or 200
(df_cpcb_dataset['cement']== 250) | (df_cpcb_dataset['cement']== 200)
```

```python
# filter dataframe based on multiple conditions
df_cpcb_dataset[(df_cpcb_dataset['cement']== 250) | (df_cpcb_dataset['cement']== 200)].value_counts('cement')
```

```python
# filter dataframe based on multiple connditions
df_cpcb_dataset[((df_cpcb_dataset['cement']== 250) | (df_cpcb_dataset['cement']== 200)) ]
```

```python
# find Cement 250 or 200 having superplastic greater then 5.0
((df_cpcb_dataset['cement']== 250) | (df_cpcb_dataset['cement']== 200)) & (df_cpcb_dataset['superplastic'] >= 5.0)
```

```python
# filter dataframe based on multiple conditions
df_cpcb_dataset[((df_cpcb_dataset['cement']== 250) | (df_cpcb_dataset['cement']== 165)) & (df_cpcb_dataset['superplastic'] >= 5.0)]
```
