```python
import pandas as pd
```

```python
# read and show the dataset
df_crop = pd.read_csv("soil_geotech_data.csv")
df_crop
```

```python
df_crop['Temp_round'] = df_crop['Temparature'].round(0)
df_crop['Moist_round'] = df_crop['Moisture'].round(0)

df_crop.pivot_table(index='Temp_round', columns='Moist_round', values='Humidity', aggfunc='mean')
```

# 1. pivot()

```python
# reshape the dataframe with .pivot()
df_crop.pivot(index="Temparature", columns="Moisture", values="Humidity")
```
