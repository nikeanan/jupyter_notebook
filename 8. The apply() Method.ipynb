# 1. apply()

```python
# use numpy functions and apply it to series
import numpy as np
df['Nitrogen'].apply(np.sqrt)
```

```python
# create your own function and apply it to a dataframe
def calculate_NPK_strength(row):
    return row['Nitrogen']+row['Phosphorous']+row['Potassium']
```

```python
df.apply(calculate_NPK_strength, axis=1)
```

```python
def evapo_rate(row):
    return row ['Temparature']/(row['Humidity']+1)
```

```python
df.apply(evapo_rate, axis=1)
```
