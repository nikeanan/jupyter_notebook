```python
import pandas as pd
```

```python
df_cpcb_dataset = pd.read_csv('concrete.csv')
```

```python
df_cpcb_dataset.head(4)
```

# 1. isin(): single Filtering

```python
# select 200 or 250 cement content
df_cpcb_dataset['cement'].isin([200,250])
```

```python
# filter dataframes
df_cpcb_dataset[df_cpcb_dataset['cement'].isin([200,250])].value_counts('cement')
```

# 2. isin(): Multiple Filtering

```python
# find 14 or 28 age from 200 or 250 cement
filter1 = df_cpcb_dataset['age'].isin([14,28])
filter2 = df_cpcb_dataset['cement'].isin([200,250])
```

```python
# filter dataframe
filter = filter1 & filter2
```

```python
df_cpcb_dataset.value_counts(filter)
```

```python
df_cpcb_dataset[filter1&filter2]
```
