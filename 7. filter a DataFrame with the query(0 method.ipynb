```python
import pandas as pd
```

```python
# read csv file
df = pd.read_csv('soil_geotech_data.csv')
# select columns
df = df[['Moisture','Crop Type','Nitrogen','Potassium','Phosphorous']]
```

```python
# show dataframe
df
```

# 1. query()

```python
# select index with moisture > 69
df.query('Moisture>69')
```

```python
# excercise : write the equivlent boolean slicing
df[df['Moisture']>69]
```

```python
# select index with moisture > 69 From Cotto
# excercise : write the equivalence boolean slicing
df.query ("Moisture>69 and 'Crop Type'=='Cotton'")
```

```python
# excercise : write the equivalence boolean slicing
df[ (df["Moisture"]>69 ) & (df['Crop Type']=='Cotton')]
```

```python
# add a not operator to the first example
df.query('not(Moisture>69)')
```

```python
# excercise: write the eauivalent boolean slicing
df[~(df['Moisture']>69)]
```

```python
# convert Nitrogen to fraction upto 1 and select those with Nitrogen> 0.4
df.query("Nitrogen/100 > 0.4")
# excercise: write the eauivalent boolean slicing
df[df["Nitrogen"]/100>0.4]
```
