# 1. Creating a conditional Column from More Than 2 Choices: np.select()

```python
import numpy as np
```

```python
# create an array based on multiple price tiers (+2 choices)
# create "conditions" and "values" variables
conditions = [
    df_cpcb_dataset['strength']>40,
    (df_cpcb_dataset['strength']>35) & (df_cpcb_dataset['strength']<=40),
    (df_cpcb_dataset['strength']>30) & (df_cpcb_dataset['strength']<=35),
    (df_cpcb_dataset['strength']>25) & (df_cpcb_dataset['strength']<=30),
    (df_cpcb_dataset['strength']>20) & (df_cpcb_dataset['strength']<=25),
    df_cpcb_dataset['strength']<=20
]
values = ['for extreme conditions','for v.severe consitions','for severe conditions','for Moderate conditions','for severe conditions',' not for R.C.C use ']
```

```python
df_cpcb_dataset['condition_category'] = np.select(conditions, values, default='unknown')
```

```python
# show dataframe
df_cpcb_dataset
```

```python
# count values inside "condition_category" column
df_cpcb_dataset.value_counts()
```

```python
# count values inside "condition_category" column
df_cpcb_dataset['condition_category'].value_counts()
```

```python
df_cpcb_dataset.value_counts('condition_category')
```

# 2. Excercise

```python
# create an array based on multiple cement content tiers (+2 choices)
# very high>500, 500>=High>400, 400>=Moderate>300, 300>=Low>200, very low<200
conditions = [
    df_cpcb_dataset['cement']>500,
    (df_cpcb_dataset['cement']>400) & (df_cpcb_dataset['cement']<=500),
    (df_cpcb_dataset['cement']>300) & (df_cpcb_dataset['cement']<=400),
    (df_cpcb_dataset['cement']>200) & (df_cpcb_dataset['cement']<=300),
    df_cpcb_dataset['cement']<200    
]

# create "conditions" and "values" variables
values  = ['Very High','High','Moderate','Low','Very Low']
values2 = ["special concrete(e.g., precast or dense waterproofing","High strength/water-resistant RCC",'RCC structural Conrete', "normal conrete(M20-M25)",'Lean conrete, weak binding']
```

```python
# set it to a new column
RCC_meaning = "Reinforced Cement Conrete"
df_cpcb_dataset['cement_content_conditions'] = np.select(conditions, values, default='  RCC_meaning')
```

```python
# show dataframe
df_cpcb_dataset
```

```python
df_cpcb_dataset.value_counts('cement_content_conditions')
```

```python
df_cpcb_dataset['cement_ RCC_meaning'] = np.select(conditions,  values2, default=' unknown')
```
