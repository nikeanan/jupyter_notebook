# 1. Create a column based on 1 conditions:
# np.where()

```python
import numpy as np
```

```python
# create as array based on strength tiers
df_cpcb_dataset['strength']>40
```

```python
np.where(df_cpcb_dataset['strength']>40,  'extremely high strength ','not for extremely severe condition')
```

```python
# set it to a new column
df_cpcb_dataset['strength_tiers'] = np.where(df_cpcb_dataset['strength']>40,  'extremely high strength ','not for extremely severe condition')
```

```python
# show first 5 rows
df_cpcb_dataset.head()
```

```python
# count values in price_tier column
df_cpcb_dataset.value_counts('strength_tiers')
```
