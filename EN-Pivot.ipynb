```python
import pandas as pd
```

```python
# read and show the dataset
df_crop = pd.read_csv("soil_geotech_data.csv")
df_crop
```

# 2.pivot_table()

```python
# make a pivot table and add an aggregete function
df_crop.pivot_table(index="Soil Type", aggfunc="sum")
```

```python
# make a pivot table, add am aggregate function and select some columns
df_crop.pivot_table(index="Soil Type", 
                    values=["Moisture", "Temparature"],
                    aggfunc="mean")
```

```python
# make a pivot table thet says how much Moisture is required on an average for the specific crop on specified soil type
df_crop.pivot_table(index="Soil Type", columns="Crop Type", values="Moisture", aggfunc="mean")
```
